```python
import random

# Definir los datos en un diccionario
data = {
    600: [170, 180, 190, 200, 210],
    850: [330, 340, 350, 360, 370],
    740: [140, 150, 160, 170, 180, 190],
    700: [160, 170, 180, 190, 200]
}

resultados_train = {}
resultados_test = {}
resultados_val = {}

# Iterar sobre cada altitud y sus respectivos RAANs
for altitud, raans in data.items():
    # Determinar la cantidad exacta para el 80% (redondeando al entero más cercano)
    cantidad_train = round(len(raans) * 0.8)
    
    # Crear una copia mezclada aleatoriamente de la lista original
    raans_mezclados = random.sample(raans, len(raans))
    
    # Dividir la lista mezclada en train y test
    resultados_train[altitud] = raans_mezclados[:cantidad_train]
    resultados_test[altitud] = raans_mezclados[cantidad_train:]

# Mostrar los resultados
for altitud in data:
    print(f"--- Altitud: {altitud} ---")
    print(f"Total RAANs : {data[altitud]}")
    print(f"Train (80%) : {resultados_train[altitud]}")
    print(f"Test (20%)  : {resultados_test[altitud]}\n")
```

```python
import random

datos_orbitales = {
    600: [170, 180, 190, 200, 210, 220],
    850: [320, 330, 340, 350, 360, 370],
    740: [140, 150, 160, 170, 180, 190],
    700: [150, 160, 170, 180, 190, 200]
}

resultados_train = {}
resultados_test = {}
resultados_val = {}

for altitud, raans in datos_orbitales.items():
    n = len(raans)
    

    # 80% for training, 10% for testing, and 10% for validation
    corte_train = int(n * 0.8)
    
    # The reaming is for testing and validation
    restantes = n - corte_train
    corte_test = corte_train + (restantes // 2) + (restantes % 2)
    
    # Shuffle the RAANs to ensure randomness
    raans_mezclados = random.sample(raans, n)
    
    # Split in the three sets
    resultados_train[altitud] = raans_mezclados[:corte_train]
    resultados_test[altitud] = raans_mezclados[corte_train:corte_test]
    resultados_val[altitud] = raans_mezclados[corte_test:]

# Mostrar los resultados
for altitud in datos_orbitales:
    print(f"--- Altitud: {altitud} (N={len(datos_orbitales[altitud])}) ---")
    print(f"Train (80%) : {resultados_train[altitud]}")
    print(f"Test (10%)  : {resultados_test[altitud]}")
    print(f"Val (10%)   : {resultados_val[altitud]}\n")
```
