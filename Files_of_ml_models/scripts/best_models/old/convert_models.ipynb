```python
import torch

from ./LoraCollisionLSTM import LoraCollisionLSTM 
from ./LoraCollisionTransformer import LoraCollisionTransformer
```

```python
model_path = 'bilstm_best_seed7.pth'

print("*"*50)
print("BILSTM MODEL")
print("*"*50)
# BiLSTM best config
#best_config = {'hidden_size': 32, 'num_layers': 3, 'dropout': 0.6, 'batch_size': 32}
best_config = {'seq_length': 8, 'hidden_size': 32, 'num_layers': 4, 'dropout': 0.3, 'batch_size': 32}
model = LoraCollisionLSTM(hidden_size=best_config['hidden_size'], num_layers=best_config['num_layers'], dropout=best_config['dropout'], add_sigmoidal=False)
weights = torch.load(model_path, map_location=torch.device('cpu'))
model.load_state_dict(weights)
model.eval()

total_params = sum(p.numel() for p in model.parameters())
print(f"Total parameters: {total_params}")

scripted_model = torch.jit.script(model)
scripted_model.save("bilstm_scripted.pt")
print("Conversion complete!")


print("*"*50)
print("TRANSFORMER MODEL")
print("*"*50)
# Transfromer best config

model_path = 'transformer_best_seed1.pth'
#best_config = {'d_model': 32, 'n_heads': 4, 'n_layers': 1, 'dropout': 0.5, 'lr': 0.001, 'batch_size': 32}
best_config = {'seq_length': 8, 'd_model': 32, 'n_heads': 4, 'n_layers': 1, 'dropout': 0.2, 'lr': 0.001, 'batch_size': 32}
model_transformer = LoraCollisionTransformer(num_numerical_features=8, d_model=best_config['d_model'], n_heads=best_config['n_heads'], n_layers=best_config['n_layers'], dropout=best_config['dropout'], add_sigmoid=False)

weights = torch.load(model_path, map_location=torch.device('cpu'))
model_transformer.load_state_dict(weights)

model_transformer.eval()
total_params = sum(p.numel() for p in model_transformer.parameters())
print(f"Total parameters: {total_params}")
scripted_model = torch.jit.script(model_transformer)
scripted_model.save("transformer_scripted.pt")
print("Conversion complete!")
```
